### Q1. WAP to find area of a circle using function. Use single return value function with argument

```python
def Area_of_Circle(r):
    return (22/7) * r * r

radius = int(input("Enter the value of radius"))
print("The Area of the circle is", Area_of_Circle(radius))
```

### Q2. WAP to find area of a triangle using function. Use single return value function with argument

```python
def Area_Of_Triangle(height, base):
    return (0.5 * base * height)

height = int(input("Enter the height of the triangle"))
base = int(input("Enter the base of the triangle"))
print(f"The area of the triangle is {Area_Of_Triangle(height,base)}")
```

### Q3. WAP to write sum, difference, product, and using arguments with return value function

```python
def calculate (a,b):
    summation = a+b
    diff = a-b
    product = a*b
    division = a/b
    mod = a%b
    return summation, diff, product, division, mod
a = int(input("Enter the 1st value"))
b = int(input("Enter the 2nd value"))
s,d,p,div,mod = calculate(a,b)
print(f"Sum = {s} | Difference = {d} | Product = {p} | Division = {div} | Modulus = {mod}")
```
