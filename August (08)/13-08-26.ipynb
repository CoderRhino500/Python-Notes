## CONDITIONAL OPERATIONS (IF, ELIF, ELSE)

```python
marks = int(input("Enter you marks"))
if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```
