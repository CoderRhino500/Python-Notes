# Inner Function

```python
def outer_function(a2, b2):
    
    def addition (a3, b3):
        return a3 + b3
        
    add = addition (a2,b2)
    return add + 5
    
result = outer_function(5, 10)
print(result)
```

## RECURSIVE FUNCTION

```python
def addition (num):
    if num:
        return num + addition (num - 1)
    else:
        return 0
res = addition (10)
print(res)
        
```

#### Q2. Create a function to find the largest of three numbers

```python
def Largest_Number(a , b, c):
    if a > b:
        if a > c:
            return f"{a} is largest"
        elif c > a:
            return f"{c} is largest"
    
    elif b > a:
        if b > c:
             return f"{b} is largest"
        elif c > b:
            return f"{c} is largest"

print("Enter Three numbers:")
a = int(input ("1st Num"))
b = int(input ("2nd Num"))
c = int(input ("3rd Num"))

print("\n")

print(Largest_Number(a,b,c))  
        
```

#### Q3. Create a function result (name, marks) and  --  if marks >= 40 return pass  -- otherwise: return fail


```python
def result(name, marks):
    if marks >= 40:
        return f"{name} has passed"
    else:
        return f"{name} has failed"

name = input("Enter Name")
marks = int(input("Enter Marks"))

result(name, marks)
```

#### Q4. Create a function **Celcius_to_fahrenheit** 
F = (C X 9/5) + 32

```python
def Celcius_To_Fahrenheit(C):
    F = (C * (9/5)) + 32
    return F

Celcius = int(input("Enter the temperature value in °C : "))
Celcius_To_Fahrenheit(Celcius)
```

#### Q5. Find the factorial of a number using recursive function

```python
# 978 is the Highest Recursion Depth
def factorial (n):
    if n == 1:
        return 1
    else:
        return n * factorial(n-1)

num = int(input("Enter a number"))
factorial(num)
```
