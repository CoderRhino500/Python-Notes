## Type Function
To verify the type of any object in python, we use **type()** function

```python
print(type(10))
```

```python
a = 11
print(type(a))
```

```python
y = 2.84
print(type(y))
```

## Boolean
Objects of boolean type may have one of two values: **True** or **False**

```python
print(type(True))
```

```python
print(type(False))
```

## Conditional Operations (IF, Elif, Else Operations)
A condition is an expression that produces either **True** or **False**

```python
# THIS IS A PROGRAM TO CHECK WHTHER YOU ARE ELIGIBLE TO VOTE OR NOT
age = int(input("Enter Age: "))
if age >= 18:
    print("You are eligible to vote")
else:
    print("Youe are NOT eligible to vote")
```

```python
marks = int(input("Enter your marks"))
if marks>= 40:
    print("Pass")
else:
    print("Fail")
```

## Comparision Operators
Greater than operator **(>)**, Lesser than operator **(<)**, Greater than or equal to Operator **(>=)**, Lesser than ot equal to operator **(<=)** assignment operator **(=)**, equal to operator **(==)**

## PRACTICE QUESTIONS

### 1. WAP to check whehter a number is positive or negative

```python
num = int(input("Enter a number"))
if num > 0 :
    print("Positive Number")
elif num < 0:
    print("Negative Number")
else:
    print("The number is 0")
```

### 2. WAP TO check whehter a number is Even or ODD

```python
num = int(input("Enter a number"))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

### 3. WAP to input two numbers and print the greater number


```python
num1 = int(input("Enter number 1: "))
num2 = int(input("Enter number 2: "))

if num1 > num2:
    print(f"{num1} is the greater number between the two")
else:
    print(f"{num2} is the greater number between the two")
```
