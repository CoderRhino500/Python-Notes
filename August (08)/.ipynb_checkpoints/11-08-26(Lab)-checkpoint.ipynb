5. Write a program to check whther a number is prime or not

```python
num = int(input("Enter a number to check if it is prime: "))
counter = 0
if num == 1:
    print("1 is neither Prime not Composite number")
else:
    for i in range(1, num+1):
        if num % i == 0:
            counter += 1
    
    if counter == 2:
        print(f"{num} is a prime number")
    else:
        print(f"{num} is a NOT prime number")
```
