## CONDITIONAL OPERATIONS (IF, ELIF, ELSE)

```python
marks = int(input("Enter you marks"))
if marks >= 80:
    print("Grade A")
elif marks >= 60:
    print("Grade B")
elif marks >= 40:
    print("Grade C")
else:
    print("Fail")
```

#### 1. WAP to input a number and check whether it is divisible by 5 or not

```python
num = int(input("Enter a number"))
if num % 5 == 0:
    print(f"{num} is divisible by 5") 
else:
    print(f"{num} is NOT divisible by 5") 
```
