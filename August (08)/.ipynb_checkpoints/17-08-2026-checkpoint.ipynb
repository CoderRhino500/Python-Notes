# VARIABLES

Variables are reserved memory location to store values. This means that when you create a variable, you reserve some space in memory. 

Now there are some rules in naming variables:
1. A variable name must start a letter or the underscore (_)
2. A variable name cannot start with number
3. A variable name can only contain alphanumeric characters and underscore (like A-Z, 0-9, and _ )
4. A variable name is case sensitive (age and Age are different)

```python
a = 100
print(a)
print(type(a))
print("\n")
b = 1000.08
print(b)
print(type(b))
print("\n")
c = "John"
print(c)
print(type(c))
```
