## Practice Question 1
Rotate a list to the right by k steps

```python
def rotate_list (lst, k):
    if not lst:
        return lst
    k = k % len(lst)
    return lst [-k :] + lst[:-k]
print (rotate_list ([1,2,3,4,5] ,2))
```

# List Loop

```python
# FOR LOOP
lst = [1,3,5,7,9]
length = len(lst)

for i in range(length):
    print(lst[i])
```

```python
# WHILE LOOP 
lst = []
length = len(lst)
i = 0
while i < length:
    print(lst[:])
    i+=1
```

# Cloning List

```python
a = [81,82,83]
b = a[:]
print(a == b)
print (a is b)
b[0] = 5
print(a)
print(b)
```

# Tuple
##### A Tuple is a collection which is ordered and unchangable. In python, tuples are written with ()

Some of the operations of tuple are:
1. Access Tuple items
2. change Tuple items
3. Loop through a tuple
4. index()
5. length()

```python
x = ('a', 'b', 'c', 'd', 'e')
print (x[2])
```

```python
x = (2, 5, 7, 8)
x[1] = 10
```

```python
x = (4, 5, 6,7,2,'aa')
for i in x:
    print(i)
```

```python
# COUNT
x = (1,2,3,4,5,6,2,10,2,11,12,2)
x.count(2)
```

```python
x = (1,2,3,4,5,6,2,10,11,12,2)
y = len(x)
print(y)
```
