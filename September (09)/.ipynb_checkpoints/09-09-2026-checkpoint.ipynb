# Break Statement

The break statement in Python is **used to exit a loop prematurely** when a specific condition is met, allowing the program to continue with the next line of code after the loop. It can be used in both for loops and while loops.

### FLowchart of Break Statement

```mermaid
graph TD;
    A[Start of Break] --> B{Test Expression};
    B -- False --> C[Exit];
    B -- True --> D{Break};
    D -- Yes --> C[Exit];
    D -- No --> E[Remaining of the body of the loop]
    E[Remaining of the body of the loop] --> B{Test Expression}
```

### Practice Question 1

WAP to display all elements before number 88

````python
for num in [50, 60, 70, 80, 88, 90, 100]:
    print (num)
    if num == 88:
        print ("Number 88 is found")
        print("Terminate Loop")
        break
    
````

### Practice Question 2
WAP to find Negative numbers from a list 

````python
num = int(input("Enter the size of the list"))
Num_list = []
print("Enter Elements")
for i in range(0, num):
    Num_list.append(int(input()))
for number in Num_list:
    if number < 0:
        print(f"Negative Number has been foud ({number}). Loop Terminated")
        break
````

# Continue Statement

The continue statement in Python is used to skip the remaining code inside a loop for the current iteration only, allowing the loop to continue with the next iteration. This means that when the continue statement is encountered, the loop does not terminate but moves on to the next cycle.

### Practice Question 3
WAP to display only odd no. 

````python
for num in [20, 11, 9, 66, 4, 89, 44]:
    if num % 2 == 0:
        continue
    print(num)
    
````

### Practice Question 4
WAP to display all positive number in a list using **"Continue"** statement

````python
num = int(input("Enter the size of the list"))
Num_List = []
print("Enter Elements")
for i in range(0, num):
    Num_List.append(int(input()))
print("The Positive numbers in list are:")
for number in Num_List: 
    if number < 0:
        continue
    print(number)
````

### Practice Question 5
WAP to display the sum of only positive numbers

````python
num = int(input("Enter the size of the list"))
Num_List = []
print("Enter Elements")
for i in range(0, num):
    Num_List.append(int(input()))
    
sum = 0
for number in Num_List: 
    if number < 0:
        continue
    sum += number

print(f"The sum only positive numbers in the list is {sum}")
````

### Practice Question 6
WAP that takes a user's desired product as an input and checks if that item is currently available in inventory

For example of OUTPUT: 
1. Enter the product you want to buy: Laptop 
   Laptop is Available

2. Enter the product you want to buy : Monitor
   Sorry, Monitor is Out of Stock
         

````python
dict = {"Laptop" : 3, "Monitor" : 4, "Mouse" : 0, "Phone" : 10, "Printer" : 0}
product = input("What product would you like to buy today?")
for key,value in dict.items():
    if key.lower() == product.lower():
        if value == 0:
            print(f"Sorry, {key} is out of Stock today")
        else:
            print(f"{key} is available. Current Stock is {value}")
````

### Practice Question 9
WAP to calculate the sum of all numbers from 1 to n

````python
def Sum_Of_N_Numbers(num):
    total = 0
    for i in range(1, num+1):
        total += i
    return total
    
num = int(input("Enter number upto which summation must be done"))
print(Sum_Of_N_Numbers(num))
````

### Practice Question 10
WAP to print multiplication table of a given number

````python
def Multiplication_Table(num, times):
    return f"{num} X {times} = {num*times}"

num = int(input("Enter a Number: "))
times = int(input("Enter the number of times it must be multiplied: "))
print("\n")

for i in range (1, times+1):
    print(Multiplication_Table(num, i))
````

### Practice Question 11

### Practice Question 13
WAP to count vowels and consonants in a sentence

````python
def Count_Vowels_And_Consonants(sentence):
    vowels, consonants = 0,0
    
    for words in sentence:
        for ch in words:
             if ch.isalpha() == True:
                if ch.lower() in "aeiou":
                    vowels += 1
                else:
                    consonants += 1
    
    return vowels, consonants

sentence = input("Enter a sentence")
vowels, consonants = Count_Vowels_And_Consonants(sentence)
print(f"There are {vowels} vowels and {consonants} consonants in the sentence.")
````
