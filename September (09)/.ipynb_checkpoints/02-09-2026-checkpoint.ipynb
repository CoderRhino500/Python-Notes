# Practice Questions

#### Q1. WAP To check whether a number is prime or not using funtion

````python
def PrimeorNot(num):
    counter = 0
    for i in range(1, num+1):
        if num % i == 0:
            counter += 1
        else:
            continue
            
    if counter == 2:
        return ("Prime")
    else:
        return ("Not Prime")
num = int(input("Enter a number"))
print("The number is ",PrimeorNot(num))
````

#### Q2. WAP To check whether a number is palindrome or not using function

````python
def isPalindrome(num):
    if str(num)[::1] == str(num)[::-1]:
        return ("Palindrome")
    else:
        return ("Not Palindrome")
number = int(input("Enter a number"))
print("The number is ", isPalindrome(number))
````

#### Q4. Write a Progrma to print first n numbers using While Loop

````python
i = 1
n = int(input("Enter a number"))

while i < n+1:
    print (i)
    i+= 1
````

### FOR LOOP

```mermaid
graph TD;
    A[Start of FOR Loop] --> B{Is condition True?};
    B -- False --> C[Body of While Loop];
    B -- True --> D[Exit to Main Body];
```

#### Syntax

for <variable> in sequence:

    statements(s)

#### FOR Loop Example

````python
numbers = [1,2,4,6,11,20]
seq = 0
for val in numbers:
    seq = val * val
    print(seq)
````

## DICTIONARY

### Iterating over a Dictionary

````python
college = {"CSE" : "Block 1", "IT" : "Block 2", "ECE" : "Block 3"}
````

````python
print("Keys are:")
for keys in college:
    print(keys)
    
````

#### Iterating over a dictionary to print values

````python
print("Values are:")
for blocks in college.values():
    print(blocks)
````

#### Iterating over a String

````python
college = "Gauhati"
for name in college:
    print(name)
````

#### Nested for Loop

for val in sequence:

    for val in sequence:

        statements(s)

#### Q5. Iterating over a string

````python
for i in range(1,6):
    for j in range(5, i-1, -1):
        print(i, end = " ")
    print('')
````
