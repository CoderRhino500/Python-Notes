### **SYNTAX TO CREATE A TEXT FILE**
variable_name = open("file.txt", file mode)

```python
f = open("hello.txt", "w+")
```

|  MODE   |       Description         |
|---------------------------------------------------------------------------------------------------|
|  "r"    |  
|  "w"    |    This mode opens file for writing. If file doesn't exist, it creates a new file |
|  "x"    |    Creates a new file                 |
|  "a"    |    Opens file in append mode          |
|  "t"    |    Opens file in text mode            |
|  "b"    |    Opens file in binary mode          |
|  "+"    |    This will open a file for reading and writing |
|----------------------------------------------------------------------------------------------------|

### Practice Question 1

##### WAP TO READ AND OPEN A FILE.

```python
a = open("hello.txt", "r")
print(a.read())
a.close()
```

#### WAP TO OPEN AND WRITE "Hello World" into a file

```python
a = open("two.txt", "w+")
a.write("Hello World")
a.close()
```

```python
a = open("two.txt", "r")
print(a.read())
a.close()
```

#### WAP TO WRITE THE CONTENT "Hi Python Program" for the existing file

```python
b = open("two.txt", "a")
b.write(" Hi Python Program")
b.close()
```
