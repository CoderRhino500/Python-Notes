## Practice Question 1
REVERSE A LIST IN PLACE (USE INPUT 1,2,3,4,5)

```python
def Reverse_List(lst, size):
    start , end = 0, size-1
    while start <= end:
        lst[start], lst[end] = lst[end], lst[start]
        start += 1
        end -= 1

    return lst

size = int(input("Enter size of the list: "))
i = 0
lst = []
for i in range(size):
    lst.append(int(input("Enter a number")))
print("Original List: ", lst)
print("Reversed List: ", Reverse_List(lst, size))
```
