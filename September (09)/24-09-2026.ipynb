## FILE MODES and their Functions

```
========================================================================================================================
|  MODE  ||     FUNCTION PERFORMED           ||                  BEHAVIOUR WITH FILE                                   |
========================================================================================================================
|  "r"    | Read only                         | Raises "FileNotFound" error if file doesn't exist.                     |
|  "r+"   | Read and Write                    | Raises "FileNotFound" error if file doesn't exist.                     |
|  "rb"   | Read in BINARY MODE               | Raises "FileNotFound" error if file doesn't exist.                     |
|  "rb+"  | Read and Write in BINARY MODE     | Raises "FileNotFound" error if file doesn't exist.                     |
|----------------------------------------------------------------------------------------------------------------------|
|  "w"    | Write only.                       | Overwrites file if it exist, otherwise creates new a new one           |
|  "w+"   | Read and Write                    | Overwrites file if it exist, otherwise creates new a new one           |
|  "wb"   | Write in BINARY MODE              | Overwrites file if it exist, otherwise creates new a new one           |
|  "wb+"  | Read and Write in BINARY MODE     | Overwrites file if it exist, otherwise creates new a new one           |
|----------------------------------------------------------------------------------------------------------------------|      
|  "a"    | Append Only.                      | Adds data to the end of existing file, otherwise creates a new one     |
|  "a+"   | Read and Append                   | Adds data to the end of existing file, otherwise creates a new one     |
|  "ab"   | Append in BINARY MODE             | Adds data to the end of existing file, otherwise creates a new one     |
|  "ab+"  | Read and Append in BINARY MODE    | Adds data to the end of existing file, otherwise creates a new one     |
|----------------------------------------------------------------------------------------------------------------------| 
|  "x"    | Exclusive Write (text)            | Raises "FileExistsError" if file already exists, else creates new file |
|  "x+"   | Exclusive Read and Write (R&W)    | Raises "FileExistsError" if file already exists, else creates new file |
|  "xb"   | Exclusive Write in BINARY MODE    | Raises "FileExistsError" if file already exists, else creates new file |
|  "xb+"  | Exclusive R&W in BINARY MODE      | Raises "FileExistsError" if file already exists, else creates new file |
|----------------------------------------------------------------------------------------------------------------------|     
```

## HOW "X" or "EXCLUSIVE WRITE" works

- To write contents into a file, we normally use "w" mode. In this mode it checks if there is any exact file name. If found, it overwrites it and if not found, then it creates a new one with that file name
  
- But the problems arises when, FOR EXAMPLE, We have a file named "data.txt" which contains crucial information. So if we use "w" mode and accidentally give the file name as "data.txt", then it will overwrite those crucial information without warning and there will be no way to recover that data

- Using "x" mode solves that. If we use "x", then it will see that a file name called "data.txt" exists and then it will raise "FileExistsError" and refuse to overwrite the crucial information on that file. And if no file with "data.txt" is detected, then it will just create a new file
