# INHERITANCE AND METHOD OVERRIDING

### Q1. Create a base class Animal() with method Speak() returning some generic sound. Create a derive class Dog that overrides speak() to return "Bark"

````python
class Animal:
    def speak(self):
        return "Some Generic Sound"

class Dog(Animal):
    def speak(self):
        return "Bark"

print(Animal().speak())
print(Dog().speak())
````

### Q2. Create a vector class taking X and Y coordinates. Overload the + operator via __add__ to perform element wise addition b/w two vectors 

````python
class Vector:
    def __init__(self, x, y):
        self.x = x 
        self.y = y
    def __add__ (self, other):
        return Vector (self.x + other.x, self.y + other.y)
    def __repr__ (self):
        return f" Vector: {self.x}, {self.y}"

v1 = Vector(2, 1)
v2 = Vector(4, 1)
print(v1 + v2)
````

### Q3. Create a class "Car" with attributes "brand" and "model". Include a method "start_engine" returning the "Brand" "Model" engine in running 

````python
class Car:
    def __init__ (self, brand, model):
        self.brand  = brand
        self.model = model
    def start_engine(self):
        return f"{self.brand} {self.model} is Running"

print("BEFORE DELETING METHOD")
Car1 = Car("Alfa Romeo", "4C")
print(Car1.start_engine())

del Car.start_engine
print("AFTER DELETING METHOD")
print(Car1.start_engine())
````

# Using Super()

### Q4. Create a base class "Person" with "name" and "age". Create subclass "Student" adding s_id using super() in its initalizer   

````python
class Person:
    def __init__(self, name, age):
             self.name = name
             self.age = age

class student(Person):
    def __init__ (self, name, age, s_id):
        super().__init__(name, age)
        self.s_id = s_id

student1 = student("Ram", 24, "S001")
print(f"{student1.name}, Age: {student1.age}, Student ID: {student1.s_id}")
````
