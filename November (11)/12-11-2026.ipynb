# Color Options in Tkinter

```python
from tkinter import *
root = tk.Tk()
root.title("Color Options in Tkinter")
button = tk.Button(root, text = "Click Me", activebackground = "blue", activeforeground =  "White")
button.pack()
label = tk.Label (root, text = "Hello, TKinter!", bg = "lightgrey", fg = "black")
label.pack()
entry = tk.Entry(root, selectbackground = "lightblue", selectforeground = "black")
entry.pack()
root.mainloop()
```

# Python Grid() Method
The **grid()** method in tkinter is ised to arrange widgets in a window using a row and column layout

```python
from tkinter import *
r = Tk()
Label( r, text = "Name").grid(row = 0, column = 0)
Entry(r). grid( row = 0, column = 1)
Label(r, text = "Age"). grid(row = 1, column = 0)
Entry(r).grid(row = 1, column = 1)
r.mainloop()
entry = tk.Entry(root,  selectbackground  = "lightblue", selectforeground = "black")
entry.pack()
root.mainloop()
```
