### Tkinter Widget


1. Label

```python
import tkinter as tk 
root = tk.Tk()
label = tk.Label(root, text = "Gauhati University")
label.pack()
root.mainloop()
```

2. Button

```python
import tkinter as tk
root = tk.Tk()
root.title("Counting Seconds")
button = tk.Button(root, text = "Stop", width = 25, command = root.destroy)
button.pack()
root.mainloop()
```
