# OVERRIDING
When inheriting from a class, we can alter the behaviour of the original superclass by "overriding" functions (i.e Declaring function in the subclass with the same name)

Functions in a subclass take Precedence over Functions a superclass (i.e SubClass Function > SuperClass Functions)

```python
class Counter:
    def __init__ (self):
        self.value = 0
    def increment(self):
        self.value += 1

class CustomCounter(Counter):
    def __init__(self,size):
        Counter.__init__(self)
        self.stepsize = size
    def increment(self):
        self.value += self.stepsize

cc = CustomCounter(4)
cc.increment() # <-- Calls increment() on CustomCounter, not Counter
print(cc.value)
```

# POLYMORPHISM
Two objects of different classes but supporting the same set of functions or attributes can be treated identically. The implementations may differ, but the outward appearance is same. 

```python
class shape:
    def __init__(self):
        pass 
        
class Rectangle(shape):
    def __init__(self, w, h):
        shape.__init__(self)
        self.width = w
        self.height = h
    def area(self):
        return self.width * self.height

class Circle(shape):
    def __init__(self,rad):
        shape.__init__(self)
        self.radius = rad
    def area(self):
        return 3.14*(self.radius**2)

l = []
l.append(Rectangle(4,5))
l.append(Circle(3))
for someshape in l:
    print (someshape.area())
```
