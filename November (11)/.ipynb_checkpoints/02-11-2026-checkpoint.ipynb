# CONSTRUCTOR
- When an instance of a class is created, the **class constructor** function is automatically called.
- The constructor always named __init__()

````python
class Person:
    def __init__(self, s):     ## Constructor func taking inital value for instance attribute value
        self.name = s
    def hello(self):
        print( self.name, "Hello")

t = Person("Jack")  # Calls __init__() on Person
t.hello()
````

### Types of constructor:

1. Default Constructor 
    - It does not take any parameters other than self

````python
class Car:
    def __init__(self):
        self.make = "Toyota"
        self.model = "Corolla"
        self.year = 2020
car = Car()
print(car.make, car.model, car.year)
````

2. Parameterized Constructor
   - It accepts arguments to initialize the object's attribute with specific values

````python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

Car = Car("Honda", "Civic", 2002)
print (f"{Car.make} {Car.model} {Car.year}")
````

# INHERITANCE
Inheritance allows us to define a class that inherits all the methods and properties from another class.

- Parent class is the class being inherited from, also called base class.

- Child class is the class that inherits from another class, also called derived class.

```

                       # THIS IS A SUPERCLASS #
                       
                            |----------|
                |---------> |  SHAPE   |  <----------|            
                |           |----------|             |

                |                 |                  |
            |----------|       |----------|     |------------|
            |  Circle  |       | Square   |     |  Rectangle |
            |----------|       |----------|     |------------|

                     ### ALL THESE ARE SUBCLASS ##
                            
```



The function & attributes of a superclass are inherited by a subclass
