## HOW TO DRAW SQUARE USING TURTLE

```python
import importlib, turtle
importlib.reload(turtle)

window = turtle.Screen()
window.bgcolor("light blue")
window.title("Turtle")

SKK = turtle.Turtle()
for i in range(4):
    SKK.forward(50)
    SKK.right(90)
turtle.done()
```

## HOW TO DRAW STAR USING TURTLE

```python
import importlib, turtle
importlib.reload(turtle)

window = turtle.Screen()
window.bgcolor("light blue")
window.title("Turtle")

star = turtle.Turtle()
star.right(75)
star.forward(100)
for i in range(4):
    star.right(144)
    star.forward(100)
turtle.done()
```

### Question 1: Write a program to draw a spiral square Outside in and Inside out

```python
import importlib, turtle
importlib.reload(turtle)

window = turtle.Screen()
window.bgcolor("light blue")
window.title("Turtle")

spsq = turtle.Turtle()
fward = 300
turn = 90
for i in range(300):
    spsq.forward(fward)
    spsq.left(turn)
    fward -= 1

turtle.done()
```

### Question 2: WAP to create spiral helix pattern
