### Types of Exceptions are:

Examples of Exception Handling

```python
a = 5
b = 0
print(a/b)
print("BYE")
```

```python
a = 5
b = 0
print(a,b)
print("bye")
b = 0
try:
    print(a/b)
except:
    print("cannot decide by zero")
```

```python
a = 5
b = 0
try:
    print(a/b)
except Exception:
    print("Number cannot be divided")
    print("Bye")
```

```python
a = 5
b = 2
try:
    print(a/b)
except Exception:
    print("Number Can't de decided")
    print("Bye")
```

```python
a = 5
b = 2
try:
    print("Resource Opened")
    print(a/b)
    print("Resource Closed")
except Exceptions as e:
    print("Number Can't de decided",e)
```

```python
a = 5
b = 0
try:
    print("Resource Opened")
    print(a/b)
    k = int(input("Enter a number"))
    print(k)
except ZeroDivisionError as e:
    print("The value cannot be decided by zero", e)
finally:
    print("Resource Closed")
```
