## Q1. Merge two dictionaries and combine matching values (Minimum 3 inputs)

```python
s1 = {"A" : 30, "B" : 10, "C" : 50, "D" : 70}
s2 = {"A" : 50, "B" : 60, "C" : 70, "D" : 80}

s3 = s1.copy()
for key,value in s2.items():
    s3[key] = s3.get(key,0) + value
print(s3)
```

## Q2. Convert two lists into dictionaries

```python
l1 = ["Name" , "Marks", "Roll Number"] ## THIS HERE IS USED AS KEY
l2 = ["Rahul", 91, 2089]               ## THIS HERE IS USED AS VALUE

## zip(<LIST-1> , <list-2>) IS THE SYNTAX
## <LIST-1> HERE WILL BE THE KEY
## <LIST-2> HERE WILL BE THE VALUE

print(dict(zip(l1,l2)))
print(dict(zip(l2,l1)))
```

## Q3. Count character frequencies in a string

```python
sentence = input("Enter Sentence")

freq = {}
for words in sentence:
    for ch in words:
        freq[ch] = freq.get(ch,0) + 1
print(freq)
```

# EXCEPTION HANDLING
- Basic Division Error Handling
- Handling Multiple Exceptions
- Using of "Else" Block
- Clean-Up with "Finally"

```python
def divison_number(s,v):
    try:
        return s/v
    except ZeroDivisionError:
        return "Cannot divide by Zero"
        
print(divison_number(5120, 5))
print(divison_number(5120, 0))
```

```python
def process_user_input( user_input ):
    try:
        num = int(user_input)
        return 100/num
    except ValueError:
        return "Invalid Num"
    except ZeroDivisionError:
        return "Error Divison by zero are not allowed"

a = input("Enter input")
process_user_input(a)
```
