```python
courses = {
    "python" : ["alice", "bob", "charlie"],
    "data structure" : ["bob", "diana"],
    "discrete data" : ["alice", "diana"]
}

inverted = {}

for course, students in courses.items():
    for student in students:
        if student not in inverted:
            inverted[student] = []
        inverted[student].append(course)
print("inverted dictionary: ")
print(inverted)
```

```python
multi_course = {student:len(course_list)
for student, course_list in inverted.items() if len(course_list)>1}

print("students enrolled in > 1 course with count: ")
print(multi_course)
```
