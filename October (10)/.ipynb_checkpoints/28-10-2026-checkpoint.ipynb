# WHAT IS OOP? 
- OOP stands for Object Oriented Programming.
- OOP is based on the Imperative Programming Paradigm, which uses statements to change a program state.
- Examples of OOP are:
  
      1. C++
      2. Java
      3. Ruby
      4. Python


- OOP uses the concept of Objects and Classes

## Pillars of OOP
1. Encapsulation
2. Inheritance
3. Polymorphism
4. Abstraction

## CORPUS

```python
class Dog:
    def __init__(self):
        pass
```

## Example Program

```python
class Dog:
    def __init__(self, arg1, arg2):  ## HERE __init__ 
        self.arg1 = arg1
        self.arg2 = arg2
    tom=Dog("tom", 12)
    print(tom.arg1)
    print(tom.arg2)
```

- Here __init__ is used to initialize the object with the the self parameter, the "arg1" and the "arg2"
- The self parameter and is a reference to the current instance of the class.

It is used to access properties and methods that belong to the class.

```python
D1 = Dog("Max", 17)
```

```python
print(D1.arg1)
print(D1.arg2)
```

## EXAMPLE PROGRAM 2

```python
class Person:
    def __init__(myobject, name, age):
        myobject.name = name
        myobject.age = age

    def greet(self):
        print("Hello, my name is " + self.name)

P1 = Person("Ram", 25)
P1.greet()
```
