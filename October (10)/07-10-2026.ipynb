## Practice Question 4

Write a modular Python function merge_and_tally(list1, list2) and do the following --

```python
list1 = [10, 50, 20, 90, 100, 200, 10]
list2 = [20, 40, 50, 10, 100]
```

1. Merge both list and counts frequency of each unique item using Python dictionaries

```python
list3 = list1 + list2
dict1 = {}
for data in list3:
    dict1[data] = dict1.get(data, 0) + 1
print(dict1)
```

2. Separates the items into two dictionaries
   - uniques : contains items that appeared exactly once
   - duplicate : contains items that appeared more than once along with their counts

```python
uniques = {}
duplicate = {}
for items, freq in dict1.items():
    if freq > 1:
        duplicate[items] = freq
    elif freq == 1:
        uniques[items] = freq
print("Unique Dictionary: ", uniques)
print("Duplicate Dictionary: ", duplicate)
```
