```python
class Person:
    def __init__ (self, name, age):
        self.name = name
        self.age = age
P1 = Person ("Ram", 26)
print(P1.name)
print(P1.age)
```

## Changing the properties of an Object

```python
P1.age = 25
print(P1.age)
```

## Deleting the properties of an object

```python
del(P1.age)
print(P1.name)
```

## Methods with Parameter


```python
class Calculator:
    def add(self, a, b):
        return a + b
    def multiply(self, a,b):
        return a*b

calc = Calculator()
print(calc.add(5,3))
print(calc.multiply(10,3))
```

## --str--() method
The --str--() method is a special method that controls what is returned when the object is printed. 
Foe example:


```python
class Person:
    def __init__(self,age,name):
        self.age = age
        self.name = name

    def __str__(self):
        return f"{self.name} ({self.age})"

    def greet(self):
        print("Hello")

P1 = Person("Shyam", 36)
print(P1)
print(P1.greet())
del Person.greet  # HERE THE GREET FUNCTION IS DELETED 
print(P1.greet()) # AND THAT IS WHY AFTER REACHING THIS STEP WE GET A "AttributeError" since there is now no greet attribute
```
